```python
nama_barang = "Buku Tulis"
harga = 15000
jumlah = 3
diskon = 2500

total = harga * jumlah
bayar = total - diskon

print("=== Struk Belanja ===")
print("Nama Barang:", nama_barang)
print("Total: Rp", int(total))
print("Diskon: Rp", int(diskon))
print("Bayar: Rp", int(bayar))
```
